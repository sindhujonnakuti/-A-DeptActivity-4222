```python
months =['Jan','Feb','Mar','Apr','May','Jun','Jul','Aug','Sep','Oct','Nov','Dec']
rainfall = [12, 18, 25, 40, 65, 120, 180, 160, 140, 80, 35, 20]

total = sum(rainfall)
average = total / len(rainfall)

wettest = months[rainfall.index(max(rainfall))]
driest = months[rainfall.index(min(rainfall))]

above_avg = [
    month for month, rain in zip(months, rainfall)
    if rain > average
]

print(f"Total rainfall: {total} mm")
print(f"Average rainfall: {average:.2f} mm")
print(f"Wettest month: {wettest}")
print(f"Driest month: {driest}")
print("Months above average:", ", ".join(above_avg))
```
