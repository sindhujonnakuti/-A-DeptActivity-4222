```python
farms = [
    {"farmer": "Ravi","crop":"ragi","area_acre":2,"yield_kg":1000},
    {"farmer": "Sita","crop":"Paddy","area_acre":3,"yield_kg":1800},
    {"farmer": "Raju","crop":"PADDY","area_acre":2,"yield_kg":1200},
    {"farmer": "Anil","crop":"Ragi","area_acre":1,"yield_kg":600},
    {"farmer": "Priya","crop":"Paddy","area_acre":2,"yield_kg":None},
    {"farmer": "Kiran","crop":"ragi","area_acre":1,"yield_kg":-50}
]

by_crop= {}
bad_records = []

for farm in farms:

    crop = farm["crop"].strip().title()
    if farm["yield_kg"] is None or farm["yield_kg"] < 0:
      bad_records.append(farm["farmer"])
      continue

    if farm["area_acre"] <= 0:
      bad_records.append(farm["farmer"])
      continue

    yield_per_acre = farm["yield_kg"]/farm["area_acre"]

    by_crop.setdefault(crop,[]).append(yield_per_acre)

    print("Bad Records:",bad_records)

    print("\nCrop-wise Farm Report:")

    for crop, yields in by_crop.items():
      number_of_farms=len(yields)
      average_yields=sum(yields)/number_of_farms

      print("Crop:",crop)
      print("Number of Farms:",number_of_farms)
      print("Average Yield per Acre:",average_yields)
      print()
```
